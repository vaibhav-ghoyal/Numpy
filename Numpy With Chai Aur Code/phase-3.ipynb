```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [Restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 230000]  # Chai Point
])

print("========== Zomato Sales Analysis ==========")
print("\n Sales Data Shape:",sales_data.shape)
print("\n Sample Data For 1st 3 Result:\n",sales_data[:3]) # 1)[0:3] 2)[:,1:] Row all but Colulmn start with 1

```

```python
# Total Sales Per Year

print(np.sum(sales_data, axis=0))

yearly_total = np.sum(sales_data[:, 1:], axis=0) # axis=0 For Column
print(yearly_total)

# Minimum Sales Per Restaurant
min_sales = np.min(sales_data[:, 1:],axis=1) # axis=1 For Row
print(min_sales)

# Maximum Sales Per Year
max_Sales = np.max(sales_data[:, 1:], axis=0)
print(max_Sales)

# Average Sales Per Restaurant 
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
# Cumulative Sales

cumsum = np.cumsum(sales_data[:,1:], axis=1)
print(cumsum)

plt .figure(figsize=(8, 6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average Cumulative Sales Accross All Restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
